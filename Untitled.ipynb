```python
# Cell DebugFit: Minimal repro of Dataset.fit → mapping
import pandas as pd
import numpy as np
from pathlib import Path
from lightfm.data import Dataset as LFDataset

# 1) Reload your Telugu DataFrame (or load from pickle if you have it)
# Option A: from pickle
# telugu_df = pd.read_pickle("telugu_df_full.pkl")

# Option B: from Parquet (uncomment if you don't have the pickle)
telugu_dir = Path("/Users/harshadayiniakula/Desktop/RS/telugu")
parquets   = list(telugu_dir.glob("*.parquet"))
telugu_df = (
    pd.concat([pd.read_parquet(p) for p in parquets], ignore_index=True)
      .drop_duplicates(subset="story_id")
      .reset_index(drop=True)
)
print("Loaded telugu_df rows:", len(telugu_df))

# 2) Reload your MIND news for verticals
mind_news_path = Path("/Users/harshadayiniakula/Desktop/RS/MINDsmall_train/news.tsv")
mind_df = pd.read_csv(
    mind_news_path, sep="\t", header=None,
    names=["newid","vertical","subvertical","title","abstract","url","te","ae"],
    dtype=str
)
verticals = sorted(mind_df["vertical"].unique())
print("Found verticals:", verticals)
print("Num verticals:", len(verticals))

# 3) Build telugu_ids
telugu_ids = [str(s) for s in telugu_df["story_id"]]
print("Num telugu_ids:", len(telugu_ids))

# 4) Fit a clean Dataset
lf = LFDataset()
lf.fit(
    users=[],               # no user metadata
    items=telugu_ids,       # your 46k article IDs
    user_features=[],       # none
    item_features=verticals  # exactly 17 verticals
)

# 5) Inspect mappings
user_id_map, item_id_map, user_feat_map, item_feat_map = lf.mapping()

print("Items registered  :", len(item_id_map),    f"(should be {len(telugu_ids)})")
print("Features registered:", len(item_feat_map), f"(should be {len(verticals)})")
```
