```python
import numpy as np, pandas as pd
from lightfm.data import Dataset as LFDataset

# Load Telugu dataset and softmax probability matrix
telugu_df = pd.read_pickle("telugu_df_full.pkl")
all_probs = np.load("telugu_all_probs_newcat.npy")
```

```python
# This is the order used during original XLM-R classification
newcat = [
    'autos','entertainment','finance','foodanddrink','health','lifestyle',
    'movies','music','news','newscrime','newspolitics',
    'newsscienceandtechnology','newstrends','newsworld','sports',
    'travel','tv','video','weather'
]

# Assign each article its most probable category
telugu_df["assigned_category"] = [newcat[i] for i in all_probs.argmax(1)]
```

```python
telugu_ids = telugu_df["story_id"].astype(str).tolist()

ds_tel = LFDataset(user_identity_features=False, item_identity_features=False)
ds_tel.fit(users=[], items=telugu_ids, item_features=newcat)

item_tuples = [
    (sid, {newcat[i]: float(p)
           for i, p in enumerate(prob_row) if p > 1e-6})
    for sid, prob_row in zip(telugu_ids, all_probs)
]

item_features_telugu = ds_tel.build_item_features(item_tuples)
item_indices = np.arange(len(telugu_ids))
```

```python
import pickle
from lightfm import LightFM

with open("lightfm_model_newcat.pkl", "rb") as f:
    model: LightFM = pickle.load(f)

print("✅ LightFM model loaded:", model)
```

```python
# LightFM stores identity‑feature rows first, then extra features.
# Your 19 vertical‑feature rows are therefore the last 19 rows.
feat_emb = model.item_embeddings[-len(newcat):]   # shape (19, dim)
```

```python
# -----------------------------------------------------------------
# Cell X : Diversified Top‑K with duplicate‑headline collapse
# -----------------------------------------------------------------
def proportional_topk(user_idx, K=10, n_top_cats=3):
    """
    Return a Top‑K recommendation list for English user `user_idx`,
    allocating items proportionally from the user’s top `n_top_cats`
    categories and collapsing duplicate headlines.
    """
    # --- 1.  User & category embeddings ---------------------------
    user_vec = model.get_user_representations()[1][user_idx]          # (dim,)
    feat_emb = model.item_embeddings[-len(newcat):]                   # (19,dim)

    # --- 2.  Category affinities & quotas -------------------------
    affinities = user_vec @ feat_emb.T                                # (19,)
    top_cats   = affinities.argsort()[-n_top_cats:][::-1]
    top_scores = affinities[top_cats]

    quotas = np.round(top_scores / top_scores.sum() * K).astype(int)
    while quotas.sum() > K: quotas[np.argmax(quotas)] -= 1
    while quotas.sum() < K: quotas[np.argmax(top_scores)] += 1

    # --- 3.  LightFM scores for *all* Telugu items ----------------
    scores = model.predict(user_idx, item_indices,
                           item_features=item_features_telugu)

    # --- 4.  Collect best items per chosen category ---------------
    rec_rows = []
    for cat_idx, q in zip(top_cats, quotas):
        cat = newcat[cat_idx]
        mask = telugu_df["assigned_category"] == cat
        if q == 0 or mask.sum() == 0:
            continue
        best_local = np.argsort(-scores[mask])[:q]
        picked = np.where(mask)[0][best_local]
        for i in picked:
            rec_rows.append((telugu_ids[i], scores[i], cat))

    # --- 5.  Build DataFrame, drop duplicate headlines ------------
    df = (
        pd.DataFrame(rec_rows, columns=["story_id", "score", "category"])
          .sort_values("score", ascending=False)
          .merge(telugu_df[["story_id", "headline"]], on="story_id", how="left")
          .drop_duplicates(subset="headline")          # collapse twins
          .reset_index(drop=True)
          .head(K)
    )
    return df
```

```python
import numpy as np, pandas as pd
rng = np.random.default_rng(42)          # reproducible randomness
sid2idx = {sid: i for i, sid in enumerate(telugu_ids)}   # map id → row index

# --- choose K random, non‑duplicate story_ids from one category ----------
def random_from_category(cat_name, k, score_array, tau=0.0, by_percentile=False):
    """
    Return up to `k` random story_ids from category `cat_name`
    that have score >= tau  (global)  OR above percentile if by_percentile=True.
    """
    idx_pool = telugu_df.index[telugu_df["assigned_category"] == cat_name]
    if len(idx_pool) == 0:
        return []
    
    pool_scores = score_array[idx_pool]

    if by_percentile:                       # percentile threshold (e.g., 70‑th)
        tau_eff = np.percentile(pool_scores, tau)   # tau now 0‑100
    else:                                   # absolute threshold
        tau_eff = tau

    good_mask = pool_scores >= tau_eff
    good_indices = idx_pool[good_mask]
    if len(good_indices) == 0:
        return []
    
    picked = rng.choice(good_indices, size=min(k, len(good_indices)), replace=False)
    return telugu_df.loc[picked, "story_id"].astype(str).tolist()
```

```python
def topk_telugu(user_idx, K=10, n_top_cats=3, mode="score"):
    """
    mode = "score"  → rank by LightFM score (current method)
    mode = "random" → pick random items in each chosen category
                      but still report their LightFM score
    """
    # 1) user‑category affinities
    user_vec = model.get_user_representations()[1][user_idx]
    cat_emb  = model.item_embeddings[-len(newcat):]          # (19, dim)
    affin    = user_vec @ cat_emb.T
    top_cats = affin.argsort()[-n_top_cats:][::-1]
    top_vals = affin[top_cats]

    # 2) slot allocation
    q = np.round(top_vals / top_vals.sum() * K).astype(int)
    while q.sum() > K: q[np.argmax(q)] -= 1
    while q.sum() < K: q[np.argmax(top_vals)] += 1

    # 3) compute LightFM score for every Telugu item once
    scores = model.predict(user_idx, item_indices,
                           item_features=item_features_telugu)

    rec_rows = []

    for cat_idx, k in zip(top_cats, q):
        cat = newcat[cat_idx]
        if k == 0:
            continue

        if mode == "random":
            picked_ids = random_from_category(
                cat, k, score_array=scores,
                tau=0.5,              # absolute threshold
                by_percentile=False   # set True and tau=70 for percentile
            )
            rec_rows += [
                (sid, float(scores[sid2idx[sid]]), cat) for sid in picked_ids
            ]

        elif mode == "score":
            mask = telugu_df["assigned_category"] == cat
            best_local = np.argsort(-scores[mask])[:k]
            picked_idx = np.where(mask)[0][best_local]
            rec_rows += [
                (telugu_ids[i], float(scores[i]), cat) for i in picked_idx
            ]

    df = (pd.DataFrame(rec_rows, columns=["story_id", "score", "category"])
          .drop_duplicates("story_id")
          .merge(telugu_df[["story_id", "headline"]], on="story_id", how="left")
          .reset_index(drop=True)
          .head(K))
    return df
```

```python
# Cell A – helpers & global data -------------------------------------------
import numpy as np, pandas as pd, matplotlib.pyplot as plt
from itertools import combinations
from scipy.spatial.distance import pdist, squareform

# assumes: telugu_df, newcat, model, item_features_telugu, item_indices, telugu_ids exist
rng   = np.random.default_rng(42)
tau   = 0  # random‑within‑category threshold
sid2idx = {sid: i for i, sid in enumerate(telugu_ids)}

def random_from_category(cat_name, k, score_array, tau=0.0, by_percentile=False):
    idx_pool = telugu_df.index[telugu_df["assigned_category"] == cat_name]
    if len(idx_pool) == 0:
        return []
    
    pool_scores = score_array[idx_pool]
    tau_eff = np.percentile(pool_scores, tau) if by_percentile else tau
    good_mask = pool_scores >= tau_eff
    good_indices = idx_pool[good_mask]
    if len(good_indices) == 0:
        return []
    
    picked = rng.choice(good_indices, size=min(k, len(good_indices)), replace=False)
    return telugu_df.loc[picked, "story_id"].astype(str).tolist()

def topk_telugu(user_idx, K=10, n_top_cats=3, mode="score"):
    user_vec = model.get_user_representations()[1][user_idx]
    cat_emb  = model.item_embeddings[-len(newcat):]  # (19, dim)
    affin    = user_vec @ cat_emb.T
    top_cats = affin.argsort()[-n_top_cats:][::-1]
    top_vals = affin[top_cats]

    q = np.round(top_vals / top_vals.sum() * K).astype(int)
    while q.sum() > K: q[np.argmax(q)] -= 1
    while q.sum() < K: q[np.argmax(top_vals)] += 1

    user_ids = np.repeat(user_idx, len(item_indices))  # ← Fix for LightFM.predict
    scores = model.predict(user_ids, item_indices,
                           item_features=item_features_telugu)

    rec_rows = []
    for cat_idx, k in zip(top_cats, q):
        cat = newcat[cat_idx]
        if k == 0:
            continue
        if mode == "random":
            picked_ids = random_from_category(cat, k, score_array=scores, tau=tau, by_percentile=False)
            rec_rows += [(sid, float(scores[sid2idx[sid]]), cat) for sid in picked_ids]
        elif mode == "score":
            mask = telugu_df["assigned_category"] == cat
            best_local = np.argsort(-scores[mask])[:k]
            picked_idx = np.where(mask)[0][best_local]
            rec_rows += [(telugu_ids[i], float(scores[i]), cat) for i in picked_idx]

    df = (
        pd.DataFrame(rec_rows, columns=["story_id", "score", "category"])
        .merge(telugu_df[["story_id", "headline"]], on="story_id", how="left")
        .dropna(subset=["story_id", "headline"])
        .drop_duplicates(subset=["headline"])  # drop identical headlines regardless of ID
        .reset_index(drop=True)
        .head(K)
    )
    return df, affin

```

```python
# Cell B – pick 3 representative users -------------------------------
user_profiles = []
for u in range(model.user_embeddings.shape[0]):
    cat_aff = model.get_user_representations()[1][u] @ model.item_embeddings[-len(newcat):].T
    top_cat = newcat[int(cat_aff.argmax())]
    user_profiles.append((u, top_cat, cat_aff))

df_users = pd.DataFrame(user_profiles, columns=["user_id", "top_cat", "affinities"])

rep_users = []
for target in ["sports", "entertainment", "newspolitics"]:
    rep_users.append(df_users.loc[df_users["top_cat"] == target, "user_id"].iloc[0])

print("Representative user IDs:", rep_users)
```

```python
# Cell D – Table 2: Top-10 recs for each user -----------------------
tables_score = {}
tables_rand  = {}

for u in rep_users:
    tables_score[u] = topk_telugu(u, K=10, mode="score")[0]
    tables_rand[u]  = topk_telugu(u, K=10, mode="random")[0]

for u in rep_users:
    print(f"\n=== USER {u}  (score‑ranked) ===")
    display(tables_score[u])
    print(f"\n=== USER {u}  (random τ={tau}) ===")
    display(tables_rand[u])
```

```python
# Cell E – diversity metrics over 50 random users --------------------
sample_users = rng.choice(model.user_embeddings.shape[0], size=50, replace=False)

def rec_list(user, mode):
    df,_ = topk_telugu(user, K=10, mode=mode)
    return df["story_id"].tolist(), df["category"].tolist(), df["score"].mean()

def headline_uniqueness(reclists):
    # avg pairwise Jaccard (1 - similarity)
    pairs = combinations(range(len(reclists)), 2)
    jacc = [(len(set(reclists[i]) & set(reclists[j])) /
             len(set(reclists[i]) | set(reclists[j])))
            for i,j in pairs]
    return 1 - np.mean(jacc)

def cat_entropy(catlist):
    p = pd.Series(catlist).value_counts(normalize=True).values
    return -(p*np.log2(p)).sum()

metrics = []
for mode in ["score","random"]:
    recs   = []
    cats   = []
    scores = []
    for u in sample_users:
        r,c,avg = rec_list(u, mode)
        recs.append(r)
        cats.append(cat_entropy(c))
        scores.append(avg)
    metrics.append([
        mode,
        headline_uniqueness(recs),
        np.mean(cats),
        np.mean(scores)
    ])

table_metrics = pd.DataFrame(metrics,
    columns=["mode","headline_uniqueness","cat_entropy","mean_score"])
display(table_metrics)
```

```python
# Cell – Overlap heatmap of Top‑10 random Telugu recommendations

import matplotlib.pyplot as plt
import seaborn as sns
from itertools import combinations
from sklearn.metrics import jaccard_score

# Step 1: Choose 10 random users
user_sample = rng.choice(model.user_embeddings.shape[0], size=10, replace=False)

# Step 2: Collect their Top‑10 random recommendations
top10_random = []
for u in user_sample:
    df = topk_telugu(user_idx=u, K=10, mode="score")[0]

    top10_random.append(set(df["story_id"]))

# Step 3: Compute 10×10 Jaccard similarity matrix
jac_matrix = np.zeros((10, 10))
for i in range(10):
    for j in range(10):
        a, b = top10_random[i], top10_random[j]
        jac_matrix[i, j] = len(a & b) / len(a | b) if len(a | b) > 0 else 0

# Step 4: Plot the heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(jac_matrix, annot=True, fmt=".2f", cmap="coolwarm",
            xticklabels=[f"U{i}" for i in range(10)],
            yticklabels=[f"U{i}" for i in range(10)])
plt.title("Jaccard Overlap – Top‑10 Score Telugu Recommendations")
plt.xlabel("User")
plt.ylabel("User")
plt.tight_layout()
plt.show()
```

```python
# Fast scan for users with at least one positive Telugu score
import numpy as np

U = model.user_embeddings                      # (n_users, d)
V = model.item_embeddings[item_indices]        # (n_telugu, d)

batch = 500                                    # tune to your RAM/GPU
good_users = []

for start in range(0, U.shape[0], batch):
    U_batch = U[start : start+batch]           # view, no copy
    # (batch, d) @ (d, n_tel) -> (batch, n_tel)
    max_scores = (U_batch @ V.T).max(axis=1)   # vectorised
    pos_mask   = max_scores > 0
    if pos_mask.any():
        rows    = np.nonzero(pos_mask)[0]
        good_ids = (start + rows).tolist()
        good_vals = max_scores[pos_mask].tolist()
        good_users.extend(zip(good_ids, good_vals))

# sort by best positive score descending
good_users.sort(key=lambda x: -x[1])
print(f"Found {len(good_users)} users with >0 Telugu score.")
print("Top 5 candidate user IDs:", [u for u,_ in good_users[:5]])
```
